## Numpy中的数据运算

相同尺寸的数组可以直接进行四则运算

```python
a = np.random.rand(2,3)
b = np.random.rand(2,3)
a, b, a+b, a-b, a*b, a/b, a//b, a%b  # 对应元素进行运算
```

np.dot实现点积运算；@实现矩阵乘法

```python
a = np.random.randint(0, 6, size=(3,3))
b = np.random.randint(0, 6, size=(3,3))
a, b, np.dot(a, b), a@b, np.matmul(a, b), a@b == np.matmul(a, b)
```

np.sqrt实现求平方根
np.sin, np.cos实现三角运算
np.log，np.power实现对数和指数运算

```python
a = np.random.randint(1, 5, size=(1, 3))
a, np.sqrt(a), np.sin(a), np.cos(a)
```

```python
np.log(a), np.power(a, 2)
```

如果把narry和一个数进行四则运算，那么所有元素都会进行四则运算，这是广播机制

四则运算永远是对应元素之间的计算，和矩阵乘法要区分开

```python
a = np.random.randint(1, 30, size=(2,3))
a, a*5, a+5, a-5, a/5, a//5, a%5
```

广播的另一个特点：如果两个数组的shape不一致，会把数组放缩为一致的shape，然后再运算

广播的前提：
NumPy 在判断两个数组是否兼容时，会比较它们的每一个维度（从最右边的维度开始往前比）。
两个维度兼容的条件只有两个：

该维度的大小相等。

其中一个维度的大小为 1。

```python
a = np.array([1,2,3])
b = np.arange(start=4,stop=7).reshape((3,1))
a, b, a+b, a*b, a/b, a//b, a%b
```

```python
# 那如果是2*3和3*2呢，会直接报错
a = np.arange(start=1, stop=7).reshape((2,3))
b = np.arange(start=-6,stop=0).reshape((3,2))
a, b, a+b, a-b, a*b, a/b, a//b, a%b
```

## min, max, argmin, argmax操作

如果都不加axis参数，就会把一个多维数组降维为一维后，从全局视角出发返回最值/最值的索引

如果加axis参数，相当于消去当前参数对应的维度，比如假设原始为(2,3)矩阵，加上axis=0，就消去了0号维度，变成(3,)，此时返回的最值/最值索引就是一行三列元素；如果axis=1，消去了1号维度，变成(2,)，此时返回的最值/最值索引就是一行两列元素。

```python
a, b, a.min(), b.min(), a.max(), b.max()
```

```python
a.min(axis=0), a.min(axis=1)
```

```python
a.argmax(), a.argmax(axis=0), a.argmax(axis=1), a.argmin(), a.argmin(axis=0), a.argmin(axis=1)
```

```python
a = np.random.randint(1, 100, size=(2,3))
```

```python
a, a.argmax(), a.argmax(axis=0), a.argmax(axis=1), a.argmin(), a.argmin(axis=0), a.argmin(axis=1)
```

 axis=1视角下，每一行的最大元素索引分别为0，0；每一行的最小索引分别为2，1

求和  a.sum([axis=])

均值 a.mean([axis=]);np.median(a, [axis=])

方差 a.var([axis=])

标准差 a.std([axis=])

```python
a, a.sum(), a.sum(axis=0), a.sum(axis=1), a.mean(), a.mean(axis=0), a.mean(axis=1)
```

```python
np.median(a, axis=0)
```

```python
a, a.var(), a.var(axis=0), a.std(), a.std(axis=0)
```

使用下标访问元素a[i, j], 也可以用原始的list方式访问a[i][j]

通过条件筛选指定的元素a[a<3]，行优先，最终返回一个一维数组

```python
a, a[0,1], a[0][1], a[a<40]
```

```python
# 可以使用逻辑运算符组合不同的条件
# 区分原生的与或操作，这里用单一单符号
a[ (a<40) & (a%2==1)], a[(a<40)|(a%2==0)]
```

切片操作，和原生的list用法一致，但是可以在一个方括号里面用逗号分隔

```python
a[0][0:2], a[0, 0:2]
```

```python
a[0], a[0, :]
```

```python
# 冒号后面可以再来一个冒号，表明跨度stride
a[0, ::2]
```

```python
a[0, ::-1], a[0, ::-2]
```

```python
a, a[::-1]
```
